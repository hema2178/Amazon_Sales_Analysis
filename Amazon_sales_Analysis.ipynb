##### Task 4 — Convert Date Column into Datetime Format

#### PART 3 — DATA PROCESSING

##### Task 1 — Create Revenue Column

```python
df['Revenue'] = (
    df['Quantity'] * df['Unit_Price']
)
df.head()
```

##### Task 2 — Add Revenue Column to Dataset

```python
df.head()
```

##### Task 3 — Check Revenue Column

```python
print(df[['Quantity', 'Unit_Price', 'Revenue']].head())
```

#### PART 4 — DATA ANALYSIS

##### Task 1 — Total Revenue

```python
total_revenue = df['Revenue'].sum()
print("Total Revenue:", total_revenue)
```

##### Task 2 — Best Selling Product

```python
product_quantity = df.groupby('Product')['Quantity'].sum()

print(product_quantity)
```

##### Task 3 — Best Performing City

```python
city_revenue = df.groupby('City')['Revenue'].sum()
print(city_revenue)
```

##### Task 4 — Category Analysis

```python
category_revenue = df.groupby('Category')['Revenue'].sum()
print(category_revenue)
```

##### Task 5 — Monthly Sales Analysis

```python
df['Month'] = df['Order_Date'].dt.to_period('M')

monthly_revenue = df.groupby('Month')['Revenue'].sum()

print("Monthly Revenue:")
print(monthly_revenue)
```

#### PART 5 — DATA VISUALIZATION

##### Task 1 — Bar Chart: Revenue by City

```python
plt.figure(figsize=(8, 5))

plt.bar(city_revenue.index, city_revenue.values)

plt.title('Revenue by City')
plt.xlabel('City')
plt.ylabel('Revenue')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

##### Task 2 — Pie Chart: Category Contribution

```python
plt.figure(figsize=(7, 7))

plt.pie(
    category_revenue.values,
    labels=category_revenue.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Category Contribution to Total Revenue')

plt.tight_layout()
plt.show()
```

##### Task 3 — Line Chart: Monthly Sales Trend

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
